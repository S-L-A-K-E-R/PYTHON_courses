## What is a Notebook, and how does it work?

A Notebook is divided into **cells**.

There are mainly two types of cells:

* **Code cells**, where you write and execute Python code.
* **Text cells**, where you can write explanations, titles, instructions, notes, equations, links, etc.

What you are reading right now... is a **text cell**!

<br>

> Text cells are very useful for organizing a Notebook into different sections and explaining what each part of your code is supposed to do.

### What is Markdown?

Text Cells use a formatting language called **Markdown**.\
Markdown is not a programming language: it is a **lightweight markup language** used to format text in a simple and readable way.

It is extremely common on the Internet. You can find Markdown in places such as:

* GitHub `README.md` files
* technical documentation
* forums and messaging platforms
* Jupyter Notebooks
* Google Colab

<br>

Instead of using buttons to format everything, you write small symbols directly into the text.

<br>

For example you can create different **levels of headings**.

```text
# Main title
## Subtitle
### Smaller subtitle
```

<br>


#### **Formatting in Markdown**

![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/COLAB_editing.png)

You can format text:

```text
**bold text**
*italic text*
`code`
```

which gives:

+ **bold text**
+ *italic text*
+ `code`

<br>

You can also create code snippet using ` ``` ` surrounding your code in the cell:
```python

```
Note that you cannot run code in *Text Cells*.


#### **Lists in Markdown**

You can easily create unordered lists using `-`, `+`, or `*`:

```text

```

or numbered lists:

```text

```

<br>

You can also create clickable [links](https://google.com/):


The structure is:


## Running code cells



![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/PYTHON_banner_chapter.png)


Now, it's time to get into Python programming.

Try running the **Code Cell** below:

````python
print("This is a Code Cell!")
print("You can execute me with the arrow on the left.")

print("The output of the code should display right below this line.")
````

Unlike a traditional program, you do not necessarily have to execute the entire Notebook at once.\
You can run **one cell at a time**, whenever you want.

When you execute a cell, only the code contained inside that cell is executed.

<br>

For example, you could first run a cell containing:


````python
# Creating a variable:
my_number = 42
````

And then, even in other code cells, you can call this newly created variable for the whole notebook.

````python
# Using the variable in a new cell:
print("Value of the variable:")
print(my_number)
````

The second cell will still know the value of `my_number`.

<br>

This is because Colab keeps a **running Python session** in memory.

Variables, imported libraries, functions, classes, and other objects you create remain available until the runtime is restarted or disconnected.

<br>

At any time on your notebook you can see what **variables are actually active** when you executed code.

At the bottom of the notebook, down-left of the screen, you can click on the `{} Variables` icon.

![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/COLAB_bottom.png)

This will open a new tab on the notebook:

![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/COLAB_variable-table.png)

⚠️ Because of this, **the order in which you execute cells matters!**

Sometimes a Notebook may appear to work simply because you previously executed another cell. This is one of the important things to keep in mind when working with notebooks.

## Running the entire Notebook

If you prefer something closer to the usual programming workflow, you can also execute **all the cells from top to bottom**.

Google Colab provides an option to run the entire Notebook automatically.

![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/COLAB_runtime-all.png)

<br>

This is also a good way to check that your Notebook really works from a clean beginning and does not accidentally depend on cells you executed earlier.

<br>

⚠️ **Before submitting your work:** restart the runtime and run every cell from top to bottom.\
Your Notebook must work without relying on anything previously stored in memory.

<br>

---


## Where is the code actually running?

When using Google Colab, your Python code is usually **not running directly on your computer**.

Google provides you with a remote runtime: essentially, a temporary computer hosted on Google's servers that executes your Python code.

During your session, this runtime keeps things such as:

* your variables,
* your imported libraries,
* your functions and classes,
* and temporary files created during execution.

However, keep in mind that this environment is **temporary**.

If the runtime is restarted or disconnected, its memory and temporary files may disappear. Important files should therefore be saved somewhere persistent, such as Google Drive or your own computer.

# Running Notebooks locally: Jupyter

![image.png](https://raw.githubusercontent.com/S-L-A-K-E-R/PYTHON_courses/main/COURSES/001_Intro/ASSETS/JNB_banner.png)

Google Colab is based on the same general Notebook philosophy as another very popular tool:

**Jupyter Notebook**

<br>

Jupyter can be installed and run directly on your computer.

Interestingly, even though it runs locally, its interface usually opens inside your **web browser** using a local address.\
The experience is therefore quite similar to Google Colab.

<br>

The main difference is where your code runs:

**Google Colab → Google's remote servers**\
**Jupyter Notebook → your own computer (locally run)**

<br>

Both approaches have advantages.

Colab is extremely convenient because there is almost nothing to install, while running Python locally gives you more control over your environment, files, libraries, and computing resources.
