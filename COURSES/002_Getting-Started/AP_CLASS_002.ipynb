## COMMENTS in Python

Adding comments to your code is extremely useful for every other programmers that might stumble upon your code... or even for your future self, because you will definitely forget your own code 3 days after writing it.

Comments are used to keep small or medium documentation about the behavior of your code.
They are really useful for explaning hard snippet of code that aren't easy to understand at first glimpse.

### Single line comment

In Python, you can use `#` to comment small snippet of code.

````python
# A Python single-line comment is done using "#"

print("hey!") # A comment can be put anywhere! But everything after the #... won't be read.

#print("Test?") 
# ^ This won't run, because the "#" is in front of the command.
````

### Multi-line comments:


For having a whole area of text, for describing functions, or adding more texts to your source code...\
You can use the quote symbol like so:

``` Python
# v these " quotes.
"""
Comment line N.1
Line N.2

(...)

I could keep going, until I reach the latest quote (").

"""
# ^ these " quotes.

```

````python
"""
If you want to have multiple lines of comments...
You can do so using the quote (") character *three times* (x2). 

The text inside the (") won't be read by the program.
"""
````

## Variables

Variables are what we use just like in maths. A variable can take whatever value, and we will use it for further uses.

Ideally, variables are set with a value from the user, so it allows infinite uses with our program.


What might be different from other programming languages:

> In Python, when you're declaring a variable there is NO NEED to set its type.\
> Python automatically will decide what is the best type to use for it.

````python
# ID: an integer (int) value (id ∈ N)
id = 4

# Name: a string (a list of characters)
message = "Hello, I'm currently learning Python!" 
````

For those who really love stricter programming language (like C, Rust...), you can set the type of a variable. It's a "**TYPING**".

If you really want to decide of a set type for a variable, you can follow this:

``` text
variable_name: TYPE
```

````python
# ID_book: an unique integer number for a book (∈ N)
ID_book: int = 502

````
