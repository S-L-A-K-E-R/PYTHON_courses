For most of your work on these notebooks (or on your .py files), you are going to use a "REPL" structure:

1. **<ins>R</ins>ead**: get some data... most of the time `FROM THE USER`
2. **<ins>E</ins>valuate**: do some calculation, conditionnal operations... whatever, that's the whole process of the algorithm!
3. **<ins>P</ins>rint**: display the result to the user.
4. **<ins>L</ins>oop**: repeat the whole operation.


<br>

---



Variables are what we use just like in maths. A variable can take whatever value, and we will use it for further uses.

Ideally, variables are set with a value from the user, so it allows infinite uses with our program.


What might be different from other programming languages:

> In Python, when you're declaring a variable there is NO NEED to set its type.\
> Python automatically will decide what is the best type to use for it.

````python
# ID: an integer value (id ∈ N)
ID = 4




````

For those who really love stricter programming language (like C, Rust...), you can set the type of a variable. It's a "**TYPING**".

If you really want to decide of a set type for a variable, you can follow this:

``` text
variable_name: TYPE
```

````python
# ID_book: an unique integer number for a book (∈ N)
ID_book: int = 502

````
