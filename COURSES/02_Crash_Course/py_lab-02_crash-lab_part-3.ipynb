# K. Errors & simple verification

Errors are useful information.

Don't train yourself to ignore the red text — **read it**.

Let's deliberately produce one:


````python
# Intentional runtime error: (#GA-IC)

number = int("definitely not a number")
print(number)
````

**Q.1)** What exception do you get?

**Q.2)** Which line caused it?

**Q.3)** Why?

A very basic way to react to an expected conversion error is:


````python
raw_value = "42"

try:
    number = int(raw_value)
    print("Converted:", number)

except ValueError:
    print("This value cannot be converted to an integer.")
````

Change `raw_value` to something invalid and observe the difference.

Don't use `try/except` to hide every error.  
Use it when an error is **expected and meaningful to handle**.
