## 2/ Conditions

Eventually, let's get into `CONTROL FLOW`: controlling what lines your program runs.

Conditions should already be familiar to you.  
What changes mostly is the **syntax**... and a few very convenient Python tools.


Here is a usual `if` in `{C}`:

```c
if (x == 1 || x == 3 || x == 7) {
    // (...)
}
else {
    if (x == 2 && y == 2) {
        // (...)
    }
}
```

Python strips away the `()`, `{}` and semicolons.

However, compared to `{C}`, you **must respect indentation**.  
Indentation defines which instructions belong to the condition.

```python
if condition_1:
    # inside IF
    ...

elif condition_2:
    # inside ELIF
    ...

else:
    # inside ELSE
    ...

# outside the condition
...
```

`elif` simply means **else if**.

### Logical operators

Python uses readable words instead of `{C}`'s `&&`, `||` and `!`:

| `{C}` | Python |
| :---: | :---: |
| `&&` | `and` |
| `\|\|` | `or` |
| `!` | `not` |

Try to predict the output **before** running the next cell.


````python
x = 7
y = 2

print( (x > 5) and  (y == 2) )
print( (x < 5) or   (y == 2) )
print( not (x == 7) )
````

### Python can make some conditions much cleaner

Instead of:

```python
if x == 1 or x == 3 or x == 7:
    ...
```

you can check whether a value is **inside a collection**:

```python
if x in (1, 3, 7):
    ...
```

This also works with **String values**!

```python
word = "Hello"

if word in "Hello World!":
    ...
```


You can also chain comparisons:

```python
if 0 <= grade <= 20:
    ...
```

In C, we would have to write:
```c
if (0 <= grade && grade <= 20){
    ...
}
    
``` 

Much nicer, right?

# C. Loops — Stop writing `{C}` loops in Python

You already know what a loop is.

The important part now is learning **how Python likes to loop**.

In `{C}`, you're very used to manipulating indexes:

```c
for (int i = 0; i < size; i++) {
    printf("%d", values[i]);
}
```

You *can* recreate something similar with `range()`:

```python
for i in range(len(values)):
    print(values[i])
```

But most of the time, Python can simply iterate over the values directly:

```python
for value in values:
    print(value)
```
