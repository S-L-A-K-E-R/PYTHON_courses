# F. Mutability, references & aliasing

Warning, there might be some trap ahead (meaning unexpected behavior).

Predict what gets displayed **before running the cell**:


````python
a = [1, 2, 3]
b = a

b.append(4)

print("A:", a)
print("B:", b)
````

**Q.1)** Were you expecting `a` to change too?

`a` and `b` don't contain two independent lists here.  
They both refer to the **same list object**.

Test their identity:


````python
print(a == b)
print(a is b)
````

Now compare that with a copy:


````python
a = [1, 2, 3]
b = a.copy()

b.append(4)

print("A:", a)
print("B:", b)

print(a == b)
print(a is b)
````

For now, remember this simple distinction:

**Mutable:**  
`list`, `dict`, `set`

**Immutable:**  
numbers, `bool`, `str`, `tuple`, `None`

# I. Comprehensions

Python can build collections directly from loops.

Instead of:

```python
squares = []

for x in range(10):
    squares.append(x ** 2)
```

you can write:

```python
squares = [x ** 2 for x in range(10)]
```

This is called a **list comprehension**.
